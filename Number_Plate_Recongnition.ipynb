```python

# Using a try-except block to avoid re-installing every time
try:
    import easyocr
except ImportError:
    !pip install easyocr

import cv2
import sqlite3
import datetime
from ultralytics import YOLO
import re
from google.colab.patches import cv2_imshow


# Defining the path to your trained model on Google Drive
MODEL_PATH = '/content/drive/MyDrive/yolo_results/license_plate_run1/weights/best.pt'

# Load your trained YOLOv8 model
print("Loading YOLOv8 model...")
model = YOLO(MODEL_PATH)
print("Model loaded.")

# Initialize the EasyOCR reader
print("Initializing EasyOCR reader...")
reader = easyocr.Reader(['en'])
print("EasyOCR reader initialized.")

# Connect to the SQLite database
DB_PATH = 'plates_database.db'
conn = sqlite3.connect(DB_PATH)
cursor = conn.cursor()

cursor.execute('''
    CREATE TABLE IF NOT EXISTS detected_plates (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        timestamp TEXT NOT NULL,
        plate_text TEXT NOT NULL
    )
''')
conn.commit()
print(f"Database setup complete at '{DB_PATH}'.")



def process_and_extract_text(image_path):
    """
    This function takes an image path, detects license plates,
    extracts text using OCR, saves it to the database,
    and displays the result.
    """
    print(f"\nProcessing image: {image_path}")
    image = cv2.imread(image_path)
    if image is None:
        print(f"Error: Could not open or find the image at {image_path}")
        return

    # Run YOLOv8 detection on the image
    results = model(image)[0]

    # Loop through each detected bounding box
    for result in results.boxes.data.tolist():
        x1, y1, x2, y2, score, class_id = result

        # Set a confidence threshold
        if score > 0.6:
            plate_crop = image[int(y1):int(y2), int(x1):int(x2)]

            ocr_result = reader.readtext(plate_crop, detail=0, paragraph=False)

            if ocr_result:
                # Clean the recognized text
                raw_text = ocr_result[0]
                # Keep only uppercase letters and numbers
                plate_text = re.sub(r'[^A-Z0-9]', '', raw_text).upper()

                if plate_text:
                    print(f"  > Detected Plate Text: {plate_text} (Confidence: {score:.2f})")

                    # Save the cleaned plate text to the database with a timestamp
                    timestamp = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")
                    cursor.execute("INSERT INTO detected_plates (timestamp, plate_text) VALUES (?, ?)", (timestamp, plate_text))
                    conn.commit()
                    print(f"  > Saved '{plate_text}' to the database.")

                    # Draw the bounding box and text on the image for visualization
                    cv2.rectangle(image, (int(x1), int(y1)), (int(x2), int(y2)), (0, 255, 0), 2)
                    cv2.putText(image, plate_text, (int(x1), int(y1) - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.9, (36, 255, 12), 2)

    # Display the final image with detections in the Colab output
    print("Displaying final annotated image.")
    cv2_imshow(image)
```

```python

try:
    import easyocr
    print("Libraries are already installed.")
except ImportError:
    print("Installing easyocr...")
    !pip install easyocr -q
    print("Installation complete.")

import cv2
import sqlite3
import datetime
from ultralytics import YOLO
import re
from google.colab.patches import cv2_imshow


# Define the path to your trained model on Google Drive
MODEL_PATH = '/content/drive/MyDrive/yolo_results/license_plate_run1/weights/best.pt'

# Load your trained YOLOv8 model
print("Loading YOLOv8 model from Google Drive...")
model = YOLO(MODEL_PATH)
print("✅ Model loaded successfully.")

# Initialize the EasyOCR reader for English
print("Initializing EasyOCR reader...")
reader = easyocr.Reader(['en'])
print("✅ EasyOCR reader initialized.")

# Connect to the SQLite database (it will be created if it doesn't exist)
DB_PATH = '/tmp/plates_database.db' # Changed database path to /tmp/
conn = sqlite3.connect(DB_PATH)
cursor = conn.cursor()

# Drop the table if it already exists to ensure the correct schema
cursor.execute("DROP TABLE IF EXISTS detected_plates")

# Create the table to store plate data, including the 'frame_id'
cursor.execute('''
    CREATE TABLE IF NOT EXISTS detected_plates (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        timestamp TEXT NOT NULL,
        frame_id INTEGER NOT NULL,
        plate_text TEXT NOT NULL
    )
''')
conn.commit()
print(f"✅ Database setup complete. Ready to save data to '{DB_PATH}'.")


# === STEP 3: DEFINE THE MAIN VIDEO PROCESSING FUNCTION ===
# This is the core function that will loop through the video.

def process_video_and_extract_text(video_path, output_path):
    # Open the video file for reading
    cap = cv2.VideoCapture(video_path)
    if not cap.isOpened():
        print(f"--- ERROR: Could not open video file: {video_path} ---")
        return

    # Get video properties to create the output file
    frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
    fps = int(cap.get(cv2.CAP_PROP_FPS))

    # Create the video writer object to save the processed video
    fourcc = cv2.VideoWriter_fourcc(*'mp4v')
    out = cv2.VideoWriter(output_path, fourcc, fps, (frame_width, frame_height))

    print(f"\n▶️ Starting video processing for '{video_path}'...")
    frame_id = 0

    # Loop through every frame in the video
    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break  # Exit the loop if the video has ended

        # Run YOLOv8 detection on the current frame
        results = model(frame)[0]

        # Process each detection in the frame
        for result in results.boxes.data.tolist():
            x1, y1, x2, y2, score, class_id = result

            # Use a confidence threshold to filter weak detections
            if score > 0.6:
                # --- This is where we CROP the plate ---
                plate_crop = frame[int(y1):int(y2), int(x1):int(x2)]

                # --- This is where we use EasyOCR to READ the text ---
                ocr_result = reader.readtext(plate_crop, detail=0, paragraph=False)

                if ocr_result:
                    raw_text = ocr_result[0]
                    # Clean the text to keep only letters and numbers
                    plate_text = re.sub(r'[^A-Z0-9]', '', raw_text).upper()

                    if plate_text:
                        # --- This is where we SAVE to the database ---
                        timestamp = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")
                        cursor.execute("INSERT INTO detected_plates (timestamp, frame_id, plate_text) VALUES (?, ?, ?)",
                                       (timestamp, frame_id, plate_text))
                        conn.commit()

                        # Draw the results on the frame for the output video
                        cv2.rectangle(frame, (int(x1), int(y1)), (int(x2), int(y2)), (0, 255, 0), 2)
                        cv2.putText(frame, plate_text, (int(x1), int(y1) - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.9, (36, 255, 12), 2)

        # Write the (potentially modified) frame to the output video file
        out.write(frame)

        # Print a progress update every 50 frames
        if frame_id % 50 == 0:
            print(f"  Processed frame {frame_id}...")

        frame_id += 1

    # Clean up and release all resources
    cap.release()
    out.release()
    cv2.destroyAllWindows()
    conn.close()
    print(f"✅ Video processing complete! Output saved to '{output_path}'.")

# === END OF SCRIPT ===
```
